Hemos organizado estos datos sobre denuncias del [DATACRIM](https://datacrim.inei.gob.pe/panel/mapa), uniendo con data de [censo](https://censos2017.inei.gob.pe/redatam/)

```python
link='https://github.com/PoliticayGobiernoPUCP/Gob_TomaDeDecisiones/raw/refs/heads/master/distritos.gpkg'

import geopandas as gpd
crimen=gpd.read_file(link, layer='crimen')
```

```python
crimen.info()
```

- 0   DEPARTAMEN: Nombre del departamento
- 1   PROVINCIA: Nombre de la provincia
- 2   DISTRITO: Nombre del distrito
- 3   INSTITUCIO: que prepara el mapa
- 4   merger: columna para añadir otras tablas
- 5   cod: Ubigeo
- 6   denun_2015: Cantidad de Denuncias en el 2015
- 7   denun_2024: Cantidad de Denuncias en el 2024
- 8   dif2024_2015: resta 2024-2015
- 9   ratio2024_2015: division 2024/2015
- 10  ratio_grupo: si disminuye, está igual o empeora (crimen 2015 al 2014)
- 11  ratio_quadrant: quadrante (spot o atipicos)
- 12  ratio_quadrant_names: nombre del quadrante (spot o atipicos)
- 13  ratio_qt: tasa agrupada en quartiles (5)- misma cantidad de distritos por grupo
- 14  ratio_atip: tasa agrupada para detectar atipicos (grupo 0 y 5)
- 15  ratio_autoGrup: tasa agrupada en grupos homogeneos
- 16  ratio_igualInter:tasa agrupada en intervalos (5)- misma ancho de intervalo
- 17  Vivienda: Viviendas en el distrito censo 2017
- 18  Hogar: Hogares en el distrito censo 2017
- 19  Poblacion: Poblacion en el distrito censo 2017
- 20  geometry: indicaciones para representar mapa

# OBJETIVO: donde focalizar la acción.

<div class="alert alert-info" role="alert">
    
## La data en bruto
</div>

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

crimen.plot(ax=ax1,column='denun2015',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax1.set_title('denuncias 2015')

crimen.plot(ax=ax2,column='denun2024',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax2.set_title('denuncias 2024')

plt.tight_layout()
plt.show()
```

<div class="alert alert-info" role="alert">
    
## conteos o tasas?

</div>

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 6))

crimen['tasaPob']=crimen['denun2024']/crimen['Poblacion']
crimen['tasaViv']=crimen['denun2024']/crimen['Vivienda']
crimen['tasaHog']=crimen['denun2024']/crimen['Hogar']

crimen.plot(ax=ax1,column='tasaPob',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax1.set_title('Tasa:denuncias por habitantes 2024')

crimen.plot(ax=ax2,column='tasaViv',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax2.set_title('Tasa:denuncias por vivienda 2024')

crimen.plot(ax=ax3,column='tasaHog',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax3.set_title('Tasa:denuncias por hogares 2024')

plt.tight_layout()
plt.show()
```

<div class="alert alert-info" role="alert">
    
## Sintetizar periodos: diferencias o razones?
</div>

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

crimen.plot(ax=ax1,column='dif2024_2015',cmap='YlOrRd',scheme="Quantiles",
            legend=True,legend_kwds={"loc": "lower left"})
ax1.set_title('diferencias')

crimen.plot(ax=ax2,column='ratio2024_2015',cmap='YlOrRd',scheme="Quantiles",
            legend=True,legend_kwds={"loc": "lower left"})
ax2.set_title('tasas')

plt.tight_layout()
plt.show()
```
