Hemos organizado estos datos sobre denuncias del [DATACRIM](https://datacrim.inei.gob.pe/panel/mapa)

```python
link='https://github.com/PoliticayGobiernoPUCP/Gob_TomaDeDecisiones/raw/refs/heads/master/distritos.gpkg'

import geopandas as gpd
crimen=gpd.read_file(link, layer='crimen')
```

```python
crimen.info()
```

- 0   DEPARTAMEN: Nombre del departamento
- 1   PROVINCIA: Nombre de la provincia
- 2   DISTRITO: Nombre del distrito
- 3   INSTITUCIO: que prepara el mapa
- 4   merger: columna para añadir otras tablas
- 5   cod: Ubigeo
- 6   denun_2015: Cantidad de Denuncias en el 2015
- 7   denun_2024: Cantidad de Denuncias en el 2024
- 8   dif2024_2015: resta 2024-2015
- 9   tasa2024_2015: division 2024/2015
- 10  tasa_grupo: si disminuye, está igual o empeora (crimen 2015 al 2014)
- 11  outlierTasa: valores atípicos
- 12  Tasa_quadrant: quadrante (spot o atipicos)
- 13  Tasa_quadrant_names: nombre del quadrante (spot o atipicos)
- 14  geometry: indicaciones para representar mapa

## OBJETIVO: donde focalizar la acción, para luego escalarla.

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

crimen.plot(ax=ax1,column='denun2015',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax1.set_title('denuncias 2015')

crimen.plot(ax=ax2,column='denun2024',cmap='YlOrRd',scheme="Quantiles",legend=True,legend_kwds={"loc": "lower left"})
ax2.set_title('denuncias 2024')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

crimen.plot(ax=ax1,column='dif2024_2015',cmap='YlOrRd',scheme="Quantiles",
            legend=True,legend_kwds={"loc": "lower left"})
ax1.set_title('diferencias')

crimen.plot(ax=ax2,column='tasa2024_2015',cmap='YlOrRd',scheme="Quantiles",
            legend=True,legend_kwds={"loc": "lower left"})
ax2.set_title('tasas')

plt.tight_layout()
plt.show()
```

```python
crimen.explore(column='tasa2024_2015',cmap='YlOrRd',scheme="Quantiles",style_kwds={'stroke':False})
```

```python
crimen.explore(column='tasa_grupo',categorical=True,style_kwds={'stroke':False},cmap='Paired')
```

```python
crimen.explore(column='outlierTasa',categorical=True,cmap='Paired',style_kwds={'stroke':False})
```

```python
from matplotlib import colors
myColMap = colors.ListedColormap([ 'white', 'pink', 'blue', 'cyan','red'])
crimen.explore(column='Tasa_quadrant_names',categorical=True,cmap=myColMap,style_kwds={'stroke':False})
```
